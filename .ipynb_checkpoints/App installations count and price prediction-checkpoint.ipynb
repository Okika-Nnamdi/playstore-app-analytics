```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings("ignore")
%matplotlib inline
```

```python
data=data.drop(data.index[10472])
```

```python
data["Reviews"]=data["Reviews"].astype(int)
```

```python
data.info()
```

```python
data["Size"].unique()
```

```python
data["Size"].isnull().sum()
```

```python
# Convert 'M' to K-equivalent strings and remove 'k'
data["Size"] = data["Size"].str.replace("M", "000")
data["Size"] = data["Size"].str.replace("k", "")
data["Size"] = data["Size"].replace("Varies with device", np.nan)
data["Size"] = data["Size"].astype(float)
```

```python
data["Size"]
```

```python
median_size = data['Size'].median()

# Fill the NaN values with the calculated median
data['Size'].fillna(median_size, inplace=True)
```

```python
data["Size"]
```

```python
data["Installs"].unique()
```

```python
data["Price"].unique()
```

```python
chars_to_remove=['+',',','$']
cols_to_clean=['Installs','Price']
for item in chars_to_remove:
    for cols in cols_to_clean:
        data[cols]=data[cols].str.replace(item,'')
```

```python
data["Price"].unique()
```

```python
data["Installs"].unique()
```

```python
data["Price"]=data["Price"].astype(float)
data["Installs"]=data["Installs"].astype(int)
```

```python
data.info()
```

```python
data["Last Updated"].unique()
```

```python
data["Last Updated"]=pd.to_datetime(data["Last Updated"])
data["Day"]=data["Last Updated"].dt.day
data["Month"]=data["Last Updated"].dt.month
data["Year"]=data["Last Updated"].dt.year
```

```python
data.info()
```

```python
data.head()
```

```python
data.drop(columns="Last Updated",inplace=True)
```

```python
data.head()
```

```python
data["Current Ver"].unique()
```

```python
data['Current Ver'].replace(['Varies with device', np.nan], '1.0', inplace=True)
data['Current Ver'] = data['Current Ver'].str.extract(r'^(\d+)', expand=False) 
data['Current Ver'] = pd.to_numeric(data['Current Ver'], errors='coerce')
data['Current Ver'].fillna(1.0, inplace=True)
```

```python
data["Current Ver"].unique()
```

```python
data["Android Ver"].unique()
```

```python
data['Android Ver'].replace(['Varies with device', np.nan], '4.0', inplace=True) 
data['Android Ver'] = data['Android Ver'].str.replace(' and up', '').str.split('.').str[0].astype(float)
```

```python
data["Android Ver"].unique()
```

```python
data.info()
```

```python
data.drop(columns=['App', 'Genres'], inplace=True) 
data = pd.get_dummies(data, columns=['Type', 'Content Rating'], drop_first=True)
```

```python
top_categories = data['Category'].value_counts().nlargest(15).index
data['Category_Cleaned'] = np.where(
    data['Category'].isin(top_categories), 
    data['Category'], 
    'Other'
)
data.drop(columns=['Category'], inplace=True)
data = pd.get_dummies(data, columns=['Category_Cleaned'], drop_first=True)
```

```python
data.info()
```

```python
bool_cols = data.select_dtypes(include=['bool']).columns
data[bool_cols] = data[bool_cols].astype(int).copy() 
```
